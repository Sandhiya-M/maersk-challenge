```python
import pandas as pd
from src.ml.train import time_based_split
from src.ml.features import build_delay_target

df = pd.read_csv('data/raw/shipments.csv')
status = df['status'].fillna('').astype(str).str.strip().str.upper()
valid = {'DELAYED','LATE','LATE_ARRIVAL','ARRIVAL_DELAYED','DELAY','DELIVERED','ON_TIME','ON-TIME','ON TIME','IN_TIME','ARRIVED','ON-TIME-ARRIVAL'}
df = df[status.isin(valid)].copy()
df['status'] = df['status'].fillna('').astype(str).str.strip().str.upper()
train_df, val_df, test_df = time_based_split(df)
for name, frame in [('train', train_df), ('val', val_df), ('test', test_df)]:
    y = build_delay_target(frame)
    print('---', name, '---')
    print('rows', len(frame))
    print('status counts', frame['status'].value_counts().to_dict())
    print('target counts', y.dropna().value_counts().to_dict())
    print('na rows', int(y.isna().sum()))
    print(frame[['shipment_id','status','booking_date','planned_arrival','actual_arrival']].head().to_dict(orient='records'))
```
